```python
district_avg = df.groupby('school_district', as_index=False)['less_than_hs'].mean()

top_bottom = pd.concat([
    district_avg.nlargest(10, 'less_than_hs'),
    district_avg.nsmallest(10, 'less_than_hs')
]).sort_values('less_than_hs', ascending=True)

plots.figure(figsize=(8, 10))
plots.barh(top_bottom['school_district'], top_bottom['less_than_hs'])
plots.xlabel('Less than HS (%)')
plots.title('Highest and lowest HS non-completion by district, NC')
plots.tight_layout()
```
